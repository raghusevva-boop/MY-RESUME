# Model Building

```python
# split into independent - x and dependent - y columns

x = df.drop(columns= ['Churn'])
y = df['Churn']
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y, train_size = 0.8, random_state = 42)
```

```python
print(x_train.shape, x_test.shape, y_test.shape, y_train.shape)
```

```python
from sklearn.tree import DecisionTreeClassifier
dt_model = DecisionTreeClassifier()
```

```python
dt_model.fit(x_train,y_train)
```

```python
y_pred_dt = dt_model.predict(x_test)
```

```python
y_pred_dt
```

```python
from sklearn.metrics import accuracy_score, classification_report
```

```python
dt_accuracy = accuracy_score(y_test, y_pred_dt)
print("Accuracy with Decison tree model", dt_accuracy)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier()
```

```python
rf_model.fit(x_train,y_train)
```

```python
y_pred_rf = rf_model.predict(x_test)
```

```python
y_pred_rf
```

```python
rf_accuracy = accuracy_score(y_test, y_pred_rf)
print("Accuracy with Random Forest model", rf_accuracy)
```

```python
model1 = RandomForestClassifier(n_estimators=300, oob_score =True)
```

```python
model1.fit(x_train,y_train)
```

```python
y_pred1 = model1.predict(x_test)
```

```python
accuracy_score(y_test,y_pred1)
```

```python
# taking different values of n_estimators(a same parameter) and will run a loop to see which gives best ans

n= [50, 100, 200, 300, 500, 700]

for i in n:
  model = RandomForestClassifier(n_estimators = i)
  model.fit(x_train,y_train)
  y_pred_loop = model.predict(x_test)
  a = accuracy_score(y_test,y_pred_loop)
  print(a, i)
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
rf = RandomForestClassifier(random_state = 42)
```

```python
param_grid = {
    'n_estimators':[50, 280, 300],
    'max_depth':[None, 10, 20],
    'max_features':['sqrt','log'],
    'min_samples_split':[2,5]
}

grid_search = GridSearchCV(estimator = rf,
                           param_grid = param_grid,
                           cv = 5,
                           scoring = 'accuracy')
```

```python
grid_search.fit(x_train,y_train)
```

```python
print("Best Parameters:",grid_search.best_params_)
print("Best Score:- ", grid_search.best_score_)
```
